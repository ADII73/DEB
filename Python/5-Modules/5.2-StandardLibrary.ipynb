```python
# Data Serialization

import json 
data={"Name":"Aditya","Age":23,"Aim":"PARA SF"}

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

print()

json_parsed=json.loads(json_str)
print(json_parsed)
print(type(json_parsed))
```

```python
# This if for CSV file processing

import csv

with open ('example.txt',mode ='w',newline='') as file:
    writer_tool=csv.writer(file)
    writer_tool.writerow(["Name","Age"])
    writer_tool.writerow(["Aditya",22])
    writer_tool.writerow(["Routh",23])
    writer_tool.writerow(["Dhoni",44])

with open ('example.txt',mode='r') as file:
    reader_tool=csv.reader(file)
    for row in reader_tool:
        print(row)
    
```

```python
# Regular Expression

import re

pattern= r'\d+'
text="Aditya will become 1254 PARA SF 123"

match=re.search(pattern,text)
print(match.group())
```

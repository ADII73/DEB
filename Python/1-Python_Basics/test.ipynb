```python
# Positional Arguments(when a lot of parameters are in a function , then the best practise is use (*args))
# def print_numbers(a,b,c,d,e) isntead of this many aregurmtns
def print_numbers(*args):
    for numbers in args:
        print(numbers)

print_numbers(1,2,3,4,5,6,"Aditya") # with positional arguments we can give as number arguments we want
```

```python
# Key words Argument(provide argument in the form of key value pair like dictionaries (**kwargs))

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="Aditya", age=22, country="India") # giving arguments like dictionaries
```

```python
#Temperature conversion
def temp_cov(temp,unit):
    if unit=="C":
        return temp *9/5+32
    elif unit == "F":
        return (temp-32)*5/9
    else:
        return None
    
print(temp_cov(25,"C"))
print(temp_cov(77,"F"))
```

```python
def pass_check(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password ):
        return False
        # loop each char in password
        # p
        # a
        # s
        # s
        # w
        # o
        # r
        # d
        # check each charracter is digit or not
        # p False
        # a False
        # s False
        # s False
        # w False
        # o False
        # r False
        # d False
        # any() will check atleast one value is true(password contain a digit)
        # not any()(not false) will check the password does NOT contain any number
    if not any(char.islower() for char in password ):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    
    return True

# print(pass_check("WeakPwd"))
print(pass_check("Str0ngPwd!"))
# print(pass_check("Aditya@2106"))
```

```python
def calculate_total_cost(cart):
    total_cost=0
    for item in cart:
        total_cost+=item["Price"]*item["Quantity"]

    return total_cost

cart=[
    {"name":"Apple","Price":0.5,"Quantity":4},
    {"name":"Orange","Price":1.5,"Quantity":6},
    {"name":"Guava","Price":2.5,"Quantity":7},
    {"name":"Banana","Price":0.5,"Quantity":4}
    ]

calculate_total_cost(cart)
```

```python
def factorial(n):
    if(n==0):
        return 1
    else:
        return n*factorial(n-1)
factorial(5)
# def count_frequency(file_path):
#     word_count={}
#     with open(file_path,'r') as file:
#         for line in file:
#             words=line.split()
#             for word in words:
#                 word=word.lower().strip('.,!?;:"\'')
#                 word_count[word]=word_count.get(word,0)+1

#     return word_count

# filepath ='1-Python_Basics/sample.txt'
# word_frequency=count_frequency(filepath)
# print(word_frequency)


def count_word_frequency(filepath):
    words_count={}
    with open(filepath,'r') as file:
        for line in file:
            words=line.split()
            for word in words:
                word=word.lower().strip('.,!?;:\'')
                words_count[word]=words_count.get(word,0)+1
    return words_count

filepath ='1-Python_Basics/sample.txt'
word_frequency=count_word_frequency(filepath)
print(word_frequency)
```

```python
def odd_even(num):
    if (num%2==0):
        print(f"The number {num} is even")
    else:
        print(f"The number {num} is odd")

odd_even(20) # calling function
```

```python
# Multiple parameters

def add(n,m):
    return n+m

result=add(2.3,-5.3)
print(result)
```

```python
# Default parameters(when a argument not provided while calling the function, default parameter overrides)

def greet(name="guest"):
    print(f"Hello {name}, welcome to paradise ")

greet("Aditya")
greet() # no argument is given so default parameter will take place which is created while function creation, if argument is passed then it will overide the default
greet("Dhoni")

```

```python
# Positional arguments with *args
# *args lets a function accept any number of positional arguments — not just a fixed number.
# *args collects extra positional arguments
# It stores them as a tuple

# def add(a,b,c,d) instead of giving this many parameters use *args
def add(*args):
    return sum(args)
 
add(1,2,3,4,5,9,10) # With help of this we can give as many arguments
```

```python
# Keywords Arguments (**kwargs) , it will take any number of arguments in the form of key value pair, therefor it will be treated as dictionary 

def print_details(**kwargs):
    for key , value in kwargs.items():
        print(f"{key}:{value}")

print_details(Name=["Aditya","ROuth"], Age=[35,36])
```

```python
# Combining positional and keywords arguments

def combined_details(*args,**kwargs):

    for val in args:
        print(f"Positional Arguments:{val}")
    print()
    print("Keyword Arguments")
    for key, value in kwargs.items():
        print(f"{key}:{value}")

combined_details(1,2,3,4,5,Name="Aditya", Age=35) # When providing arguments give positional arguments first then keywords arguments only else it will give error 

# Positional Arguments will come first only then keywords arguments will come this is the only format 
```

```python
# Return multiple parameters in functions

def multiply (a,b):
    return a*b,a,b

multiply(2,3)
```

```python
# You want to apply the same operation to all items in an iterable. Logic is simple You want cleaner code than a loop

# map(function,iterable):

number=[1,2,3,4,5]
def square(num):  # This function is squaring a number, i want to have same in list items, for that need to use loops insted use map
    return num*num
print(square(5))
print()

map(square,number) # this will give a map object iterator, to have the element either interate the object or convert it to list

m=map(square,number) 
for val in m:
    print(val)
print()
new=list(map(square,number))
print(new)
```

```python
# Lambda with map

numbers=[1,2,3,4]
list(map(lambda x:x*x,numbers))  # lambda takes arguments and expression
                                 # map takes a function and iterable , to execute the function for each item in the iterable
```

```python
# Multiple iterables in map()

num1=[1,2,3]
num2=[3,4,5]

added=list(map(lambda x,y:x+y, num1,num2))
print(added) #Python reads this as:“I have a function that needs 2 inputs (x, y) and I have 2 iterables (num1, num2)→ so I’ll take one value from each and pass them together”


```

```python
# List of strings to integer (can also use inbuild function in map)

str_list=["1","2","3","4"]
int_list=list(map(int,str_list)) # int function is used for typecasting
print(int_list)

words=["Aditya","Routh","sdfdsf"]
upper=list(map(str.upper,words))
print(upper)
```

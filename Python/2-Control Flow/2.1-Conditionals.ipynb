```python
## If-elif-else
age=15
if(age<13):
    print("You are child")
elif(age<18):
    print("Teenager")
else:
    print("Adult")
```

```python
## Nested condtion
Num=int(input("Enter number"))
if(Num>=0):
    print("Positive")
    if(Num%2==0):
        print("Even")
    else:
        print("Odd")
else:
    print("Negative ")

    
```

```python
# leap year

year=int(input("Enter year"))

if(year%4==0):
    if(year%100==0):
        if(year%400==0):
            print(year, "This is leap year")
        else:
            print(year,"This is not a leap year")
    else:
        print(year,"This is a leap year")
else:
    print(year,"This is not a leap year")
```

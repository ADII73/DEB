```python
## range(start,end,step)

for i in range(1,10): ## from 1 to 10
    print(i)
print()
for i in range(1,10,2): ## from 1 to 10 with jump 2  
    print(i)
```

```python
# while loop , iterates till the condtion is fulfilling  
count=0
while(count<5):
    print(count)
    count=count+1
```

```python
# break
for i in range(10):
    if(i==5):
        break ## when i becomes 5 and it matches the condition it will come out from that iteration
    print(i)
print()
# continue
for i in range(10):
    if(i%2==0):
        continue # when i=0 it meets the condition therefore it skips, when i=1 it doesnt meet the condition therfore prints  
    print(i)

# pass -  it does nothing 
```

```python
# nested loops
for i in range(3):
    for j in range(2):
        print(f"The value of i: {i} and j: {j}") # this is f string
```

```python
# sum of n natural numbers (for loops)

n=10
sum=0
for i in range(n+1):
    sum=sum+i
print(sum)

# while loop
n=10 
sum=0
count=0
while(count<=n):
    sum=sum+count
    count=count+1
print(sum)
```

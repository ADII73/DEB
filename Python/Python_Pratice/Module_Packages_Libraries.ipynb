```python
# Data Engineering Task 2 – Convert Employee Records to JSON
# Write a program that:

# 1. Imports csv and json.
# 2. Reads the employees.csv file from the previous task.
# 3. Creates an empty list named employees.
# 4. For each employee row:
#    a. Converts the salary from a string to an integer.
#    b. Adds the updated dictionary to the employees list.
# 5. Converts the employees list into a JSON string.
# 6. Prints the JSON string.
# 7. Converts the JSON string back into a Python list.
# 8. Prints the number of employees in the restored list.

# Rules:
# Use csv.DictReader() to read the file.
# Use int() to convert each salary.
# Leave id as a string.
# Use json.dumps() with indent=2 for readable JSON.
# Use json.loads() to convert the JSON string back.
# Do not hardcode the employee count.

# Hint:
# You can update a dictionary value:
# row["salary"] = int(row["salary"])
#
# Use .append() to add each dictionary to your list.
# Use len() to count the employees.

# Expected Output:
# [
#   {
#     "id": "1",
#     "name": "Aditya",
#     "department": "Engineering",
#     "salary": 30000
#   },
#   {
#     "id": "2",
#     "name": "Priya",
#     "department": "Finance",
#     "salary": 40000
#   },
#   {
#     "id": "3",
#     "name": "Rahul",
#     "department": "HR",
#     "salary": 25000
#   }
# ]
# Employee Count: 3

# Data Engineering Connection:
# CSV → Clean Data Types → JSON → Python List
# JSON is a common format used to exchange data with APIs.

import csv
import json

employee = []

with open('employees.csv', mode = 'r') as file:
    reader_tool = csv.DictReader(file)
    for row in reader_tool:
        row['salary'] = int(row['salary'])
        employee.append(row)

json_string = json.dumps(employee,indent=1)
print(json_string)
restored_employees = json.loads(json_string)
print(f"Employee Count: {len(restored_employees)}")
```

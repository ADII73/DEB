```python
# Task 1: Daily Sales Report
# A retail company gives you the sales amount for each transaction of the day.

# Your tasks:
# Calculate the total sales.
# Calculate the average sale amount.
# Find the highest and lowest sale.
# Count how many sales are greater than 1500.

sales = [1200, 800, 1500, 2000, 500, 1800, 2200]
greater_than_1500 = 0
##print(total_sales)
total_sales = 0
count = 0
for i in sales:
    total_sales = total_sales + i
    count += 1
    if i > 1500:
        greater_than_1500 += 1

print(total_sales)
average_amount = total_sales / count
print(average_amount)
print(max(sales))
print(min(sales))
print(greater_than_1500)






```

```python
# Task 2: Employee Salary Analysis
# You receive employee salaries from the HR department.

# Your tasks:
# Count employees earning more than 50000.
# Print "Bonus Eligible" for employees earning more than 60000.
# Calculate the total payroll amount.

salaries = [25000, 50000, 75000, 30000, 90000, 45000]
bonus = 15000
greater_than_50000 = 0
total_amount = 0
for ids, amount in  enumerate(salaries):
    if amount > 50000: 
        greater_than_50000 += 1
    if amount > 60000:
        print(f"Id: {ids} is eligible as salary: {amount}")
        print(f"Net salary is: {amount + bonus}")
    total_amount += amount
print(greater_than_50000)
print(total_amount)
        
```

```python
# Task 3: Server Log Monitoring
# A system administrator gives you server log statuses.

# Your tasks:
# Count the number of ERROR logs.
# Count the number of WARNING logs.
# Print a message if the number of errors exceeds 2.

logs = ["INFO", "ERROR", "ERROR" , "INFO", "WARNING", "ERROR", "ERROR", "INFO"]

count_error = 0
count_warning = 0
for logging in logs:
    if logging == 'ERROR':
        count_error += 1
    elif logging == 'WARNING':
        count_warning += 1 
    
if count_error > 2:
        print("⚠️ ALERT: ERROR logs exceeded the threshold.")
print(count_error)
print(count_warning)
```

```python
# Mini Project 1: Daily Sales Report
# Scenario
# You are a Junior Data Engineer at an e-commerce company.
# Every evening, your manager gives you a list of daily sales and asks you to generate a report.

# Your task
# Generate a report that prints:
# ----------- DAILY SALES REPORT -----------
# Total Sales:
# Average Sale:
# Highest Sale:
# Lowest Sale:
# Number of sales greater than 2000:
# Number of sales less than 1000:
# ------------------------------------------
# Rules
# Use a for loop.
# Don't use sum().
# Try not to use max() or min() (implement them yourself for practice).
# Use meaningful variable names.

sales = [1250, 3400, 1800, 900, 4500, 2100, 1600, 700, 3200]

total_sales = 0
no_of_sales = 0
Average_Sales = 0
Maximum_Sale = sales[0]
Minimum_Sale = sales[0]
Sales_Greater_2000 = 0
Sales_Less_1000 = 0
for amount in sales:
    if Maximum_Sale < amount:
        Maximum_Sale = amount

    if Minimum_Sale > amount:
        Minimum_Sale = amount
    total_sales += amount

    no_of_sales += 1
    if amount > 2000:
        Sales_Greater_2000 += 1
    elif amount < 1000:
        Sales_Less_1000 +=1

print("----------- DAILY SALES REPORT -----------")
print(f"Total Sales: {total_sales}")
Average_Sales = total_sales / no_of_sales
print(f"Average Sales: {int(Average_Sales)}")
print(f"Highest Sale: {Maximum_Sale}")
print(f"Lowest Sale: {Minimum_Sale}")
print(f"Number of sales greater than 2000: {Sales_Greater_2000}")
print(f"Number of sales less than 1000: {Sales_Less_1000}")
print("------------------------------------------")



```

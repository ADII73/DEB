```python
# Task 1 – Handle an Invalid Number
# Write a program that:

# 1. Asks the user to enter their age using input().
# 2. Tries to convert the input into an integer using int().
# 3. If conversion succeeds, prints the age.
# 4. If conversion fails, catches ValueError and prints:
#    Invalid input. Please enter a whole number.
# 5. Prints "Program finished." after the try/except blocks.

# Rules:
# Use try and except ValueError.
# Keep the conversion inside the try block.
# Do not use a loop yet.
# Do not check whether the age is positive yet.

# Test 1 – Valid Input:
# Enter your age: 25

# Expected Output:
# Your Age: 25
# Program finished.

# Test 2 – Invalid Input:
# Enter your age: apple

# Expected Output:
# Invalid input. Please enter a whole number.
# Program finished.

# Run the program separately for each test.

try:
    a = int(input("Enter your age"))
    print(a)
    print(type(a))
except ValueError:
    print("Invalid input. Please enter a whole number.")
finally:
    print("Program Finished")
```

```python
# Task 2 – Handle Division by Zero
# Write a program that:

# 1. Asks the user to enter a number.
# 2. Converts the input into an integer.
# 3. Divides 100 by that number.
# 4. Prints the result if the division succeeds.
# 5. Catches ZeroDivisionError and prints:
#    Cannot divide by zero. Please enter a nonzero number.
# 6. Prints "Program finished." after the try/except blocks.

# Rules:
# Put the conversion and division inside try.
# Use except ZeroDivisionError.
# Assume the user enters a valid integer for this task.
# Do not use a loop.
# Do not use finally yet.

# Test 1 – Valid Input:
# Enter a number: 4

# Expected Output:
# Result: 25.0
# Program finished.

# Test 2 – Zero:
# Enter a number: 0

# Expected Output:
# Cannot divide by zero. Please enter a nonzero number.
# Program finished.

# Test 3 – Negative Number:
# Enter a number: -5

# Expected Output:
# Result: -20.0
# Program finished.

# Run the program separately for each test.


try:
    num = int(input("Enter a number: " ))
    result =  100 / num
    print(f"Result: {result}")

except ZeroDivisionError:
    print("Cannot divide by zero. Please enter a nonzero number.")

finally:
    print("Program Finished")
```

```python
# Task 3 – Handle Multiple Exceptions
# Write a program that:

# 1. Asks the user to enter a number.
# 2. Converts the input into an integer.
# 3. Divides 100 by that number.
# 4. Prints the result if the calculation succeeds.
# 5. Catches ValueError and prints:
#    Invalid input. Please enter a whole number.
# 6. Catches ZeroDivisionError and prints:
#    Cannot divide by zero. Please enter a nonzero number.
# 7. Prints "Program finished." after the try/except blocks.

# Rules:
# Put the conversion and division inside one try block.
# Write two separate except blocks.
# Catch each exception by its specific name.
# Do not use a loop, else, or finally yet.

# Test 1 – Valid Input:
# Enter a number: 4

# Expected Output:
# Result: 25.0
# Program finished.

# Test 2 – Invalid Input:
# Enter a number: apple

# Expected Output:
# Invalid input. Please enter a whole number.
# Program finished.

# Test 3 – Zero:
# Enter a number: 0

# Expected Output:
# Cannot divide by zero. Please enter a nonzero number.
# Program finished.

# Run the program separately for each test.



try:
    num = int(input("Enter a number: " ))
    result =  100 / num
    print(f"Result: {result}")

except ZeroDivisionError:
    print("Cannot divide by zero. Please enter a nonzero number.")

except ValueError:
    print("Invalid input. Please enter a whole number.")

finally:
    print("Program Finished")
```

```python
# Task 4 – Use else and finally
# Write a program that:

# 1. Asks the user to enter a number.
# 2. Converts the input into an integer.
# 3. Divides 100 by that number.
# 4. Catches ValueError and prints:
#    Invalid input. Please enter a whole number.
# 5. Catches ZeroDivisionError and prints:
#    Cannot divide by zero.
# 6. Uses else to print the result only when try succeeds.
# 7. Uses finally to print:
#    Calculation attempt finished.

# Rules:
# Put only the input, conversion, and division inside try.
# Put the result print statement inside else.
# Put the finishing message inside finally.
# Use separate handlers for the two exception types.
# Do not use a loop.

# Test 1 – Valid Input:
# Enter a number: 4

# Expected Output:
# Result: 25.0
# Calculation attempt finished.

# Test 2 – Invalid Input:
# Enter a number: apple

# Expected Output:
# Invalid input. Please enter a whole number.
# Calculation attempt finished.

# Test 3 – Zero:
# Enter a number: 0

# Expected Output:
# Cannot divide by zero.
# Calculation attempt finished.

# Hint:
# Block order: try → except → except → else → finally
# else runs when try succeeds.
# finally runs whether try succeeds or an exception occurs.

# Run the program separately for each test.


try:
    num = int(input("Enter a number: " ))
    result =  100 / num

except ZeroDivisionError:
    print("Cannot divide by zero. Please enter a nonzero number.")

except ValueError:
    print("Invalid input. Please enter a whole number.")

else:
    print(f"Result: {result}")

finally:
    print("Program Finished")
```

```python
# Task 5 – Handle a Missing File
# Write a program that:

# 1. Asks the user to enter a text file's name.
# 2. Tries to open that file in read mode using with open().
# 3. Reads its contents into a variable.
# 4. Catches FileNotFoundError and prints:
#    File not found. Please check the file name.
# 5. Uses else to print the contents if reading succeeds.
# 6. Uses finally to print:
#    File reading attempt finished.

# Preparation:
# Create a file named message.txt in your notebook's working folder.
# Put this text inside it:
# Hello, Data Engineer!

# Rules:
# Put opening and reading inside try.
# Use with open() so the file closes automatically.
# Use encoding="utf-8" when opening the file.
# Put the contents print statement inside else.
# Catch FileNotFoundError specifically.
# Do not create a missing file automatically.

# Test 1 – Existing File:
# Enter a file name: message.txt

# Expected Output:
# File Contents: Hello, Data Engineer!
# File reading attempt finished.

# Test 2 – Missing File:
# Enter a file name: missing.txt

# Expected Output:
# File not found. Please check the file name.
# File reading attempt finished.

# Hint:
# file.read() returns the contents of an open file.
# finally prints the finishing message; with handles closing the file.

# Run the program separately for each test.

try:
    file_name = input("Enter file name with extension: ")
    with open (file_name) as file:
        content = file.read()
except FileNotFoundError :
    print("File not found. Please check the file name.")

else:
    print("File Contents:",content)

finally:
    print("File reading attempt finished.")
```

```python
# Data Engineering Task 1 – Handle Invalid Salary Values
# Write a program that:

# 1. Starts with these salary values:
#    
# 2. Creates an empty list named accepted_salaries.
# 3. Creates a rejected_count variable starting at 0.
# 4. Loops through salary_values.
# 5. For each value:
#    Tries to convert it into an integer.
#    If conversion succeeds, adds it to accepted_salaries.
#    If ValueError occurs, increases rejected_count by 1.
# 6. After the loop, prints:
#    The accepted salaries.
#    The number of accepted salaries.
#    The number of rejected values.
#    The total of the accepted salaries.

# Rules:
# Put try/except inside the loop.
# Catch ValueError specifically.
# Continue processing after an invalid value.
# Do not add invalid values to accepted_salaries.
# Use len() and sum() for the accepted count and total.
# Do not hardcode the calculated results.
# Only check integer conversion in this task.

# Expected Output:
# Accepted Salaries: [30000, 25000, 40000]
# Accepted Count: 3
# Rejected Count: 2
# Total Salary: 95000

# Data Engineering Connection:
# One invalid value should not prevent the remaining
# salary values from being processed.

salary_values = ["30000", "unknown", "25000", "", "40000"]
accepted_salaries = []
rejected_count = 0
accepted_count = 0

for value in salary_values:
    try:
        salary = int(value)
        accepted_salaries.append(salary)
        accepted_count +=1

    except ValueError:
        rejected_count +=1


print(f"Accepted Salaries: {accepted_salaries}")
print(f"Accepted count: {accepted_count}")
print(f"Rejected count: {rejected_count}")
print(f"Total Salaryt: {sum(accepted_salaries)}")
```

```python
# Data Engineering Task 2 – Raise Errors for Invalid Records
# Write a program that:

# 1. Starts with these employee records:

# 2. Defines a function named validate_employee(employee).
# 3. Inside the function:
#    If the name is empty after removing surrounding spaces,
#    raises ValueError with this message:
#    Employee name cannot be empty.
#
#    If the salary is zero or negative,
#    raises ValueError with this message:
#    Salary must be greater than zero.
#
#    If both checks pass, returns the employee dictionary.

# 4. Creates an empty list named accepted_employees.
# 5. Creates a rejected_count variable starting at 0.
# 6. Loops through the records and calls validate_employee()
#    inside try.
# 7. If validation succeeds:
#    Adds the returned dictionary to accepted_employees.
#    Prints "Accepted: " followed by the employee's name.
# 8. If ValueError occurs:
#    Increases rejected_count by 1.
#    Prints "Rejected: " followed by the error message.
# 9. After the loop, prints the accepted and rejected counts.

# Rules:
# Use raise ValueError(...) inside the function.
# Use except ValueError as error in the loop.
# Use .strip() to check for an empty name.
# Assume every salary is already a number.
# Use len() for the accepted count.
# Do not hardcode the counts.

# Expected Output:
# Accepted: Aditya
# Rejected: Salary must be greater than zero.
# Rejected: Salary must be greater than zero.
# Rejected: Employee name cannot be empty.
# Accepted: Neha
# Accepted Count: 2
# Rejected Count: 3

# Interview Question:
# Why doesn't Python automatically raise an exception
# when an employee's salary is -500?



employees = [
    {"name": "Aditya", "salary": 30000},
    {"name": "Priya", "salary": 0},
    {"name": "Rahul", "salary": -500},
    {"name": "   ", "salary": 25000},
    {"name": "Neha", "salary": 40000}
]

accepted_count = 0
rejected_count = 0

for i in employees:
    try:
        if i["name"].strip() == "":
            raise ValueError("Employee name cannot be empty.")

        if i["salary"] <= 0:
            raise ValueError("Salary must be greater than zero.")

        accepted_count += 1
        print(f"Accepted: {i['name']}")

    except ValueError as error:
        rejected_count += 1
        print(f"Rejected: {error}")

print(f"Accepted Count: {accepted_count}")
print(f"Rejected Count: {rejected_count}")




        

```

```python
employees = [
    {"name": "Aditya", "salary": 30000},
    {"name": "Priya", "salary": 0},
    {"name": "Rahul", "salary": -500},
    {"name": "   ", "salary": 25000},
    {"name": "Neha", "salary": 40000}
]


def validate_employee(employee):
    if employee["name"].strip() == "":
        raise ValueError("Employee name cannot be empty.")

    if employee["salary"] <= 0:
        raise ValueError("Salary must be greater than zero.")

    return employee


accepted_employees = []
rejected_count = 0

for i in employees:
    try:
        validated_employee = validate_employee(i)
        accepted_employees.append(validated_employee)
        print(f"Accepted: {i['name']}")

    except ValueError as error:
        rejected_count += 1
        print(f"Rejected: {error}")

print(f"Accepted Count: {len(accepted_employees)}")
print(f"Rejected Count: {rejected_count}")
```

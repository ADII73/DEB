```python
# Task 4 – Update an Object Through a Method

# --------------------------------------------------
# Data to Work With
# --------------------------------------------------

# Employee 1:
# Name: Aditya
# Department: Engineering
# Monthly Salary: 30000

# Employee 2:
# Name: Priya
# Department: Finance
# Monthly Salary: 40000

# Salary Raise:
# Aditya receives a 10% raise.
# Priya receives no raise.

# --------------------------------------------------
# Problem Statement
# --------------------------------------------------

# Extend your Employee class with a method named
# apply_raise(percentage).
#
# The method must increase the monthly salary stored on
# the current object by the given percentage and return
# the updated monthly salary.
#
# Apply the raise only to Aditya's existing object.
# Display both employees' salaries before and after the raise.
# Then display their annual salaries using your existing method.
#
# Keep printing outside the class.
# Do not create replacement employee objects.
# Do not directly assign the updated salary outside the class.
# Do not hardcode calculated results.

# --------------------------------------------------
# Expected Output
# --------------------------------------------------

# Before Raise:
# Aditya: 30000
# Priya: 40000
#
# After Raise:
# Aditya: 33000.0
# Priya: 40000
#
# Annual Salaries:
# Aditya: 396000.0
# Priya: 480000


class Employee:
    def __init__(self, name, department, salary):
        self.empployee_name = name
        self.employee_department = department
        self.employee_salary = salary

    def calculate_annual_salary(self):
        self.annual_salary = self.employee_salary * 12
        return self.annual_salary

    def salary_raise(self, percentage):
        self.employee_salary += self.employee_salary * percentage / 100
        return self.employee_salary

employee_1 = Employee("Aditya", "Engineering", 30000)
employee_2 = Employee("Priya", "Finance", 40000)

print(f"Before Raise: ")
print(f"{employee_1.empployee_name}: {employee_1.employee_salary}")
print(f"{employee_2.empployee_name}: {employee_2.employee_salary}")
print()
employee_1.salary_raise(10)
print(f"After Raise")
print(f"{employee_1.empployee_name}: {employee_1.employee_salary}")
print(f"{employee_2.empployee_name}: {employee_2.employee_salary}")
print()
print(f"Annual Salaries")
print(f"{employee_1.empployee_name}: {employee_1.calculate_annual_salary()}")
print(f"{employee_2.empployee_name}: {employee_2.calculate_annual_salary()}")


```

```python
# Task 5 – Default Values and Input Validation

# --------------------------------------------------
# Data to Work With
# --------------------------------------------------

# Account 1:
# Owner: Aditya
# Starting balance: Not supplied

# Account 2:
# Owner: Priya
# Starting balance: 5000

# Account 3:
# Owner: Rahul
# Starting balance: -500

# Account 4:
# Owner: "   "
# Starting balance: 1000

# --------------------------------------------------
# Problem Statement
# --------------------------------------------------

# Create a BankAccount class that stores an owner and balance.
#
# When no starting balance is supplied, it must default to 0.
#
# An empty or spaces-only owner must raise ValueError:
# Owner name cannot be empty.
#
# A negative starting balance must raise ValueError:
# Starting balance cannot be negative.
#
# A starting balance of zero is valid.
#
# Include a get_balance() method that returns the stored balance.
#
# Attempt to create all four accounts.
# Handle validation exceptions outside the class so that one
# invalid account does not prevent testing the remaining accounts.
#
# Print valid owners and their balances using object attributes
# and get_balance(). Print a rejection message for invalid accounts.
#
# For Aditya, omit the balance argument to test the default.

# --------------------------------------------------
# Expected Output
# --------------------------------------------------

# Aditya's Balance: 0
# Priya's Balance: 5000
# Invalid account: Starting balance cannot be negative.
# Invalid account: Owner name cannot be empty.

class BankAccount:
    def __init__(self, owner, balance = 0):
        if owner.strip() == "":
            raise ValueError("Owner name cannot be empty.")

        if balance < 0:
            raise ValueError("Starting balance cannot be negative.")

        self.owner_name = owner
        self.owner_balance = balance

    def get_balance(self):
        return self.owner_balance

try:
    account_1 = BankAccount("Aditya")
    print(f"{account_1.owner_name}'s Balance: {account_1.get_balance()}")

except ValueError as error:
    print(f"Invalid account: {error}")

try:
    account_2 = BankAccount("Priya", 5000)
    print(f"{account_2.owner_name}'s Balance: {account_2.get_balance()}")

except ValueError as error:
    print(f"Invalid account: {error}")

try:
    account_3 = BankAccount("Rahul", -500)
    print(f"{account_3.owner_name}'s Balance: {account_3.get_balance()}")

except ValueError as error:
    print(f"Invalid account: {error}")

try:
    account_4 = BankAccount("   ", 1000)
    print(f"{account_4.owner_name}'s Balance: {account_4.get_balance()}")

except ValueError as error:
    print(f"Invalid account: {error}")

```

```python
# Interview Task 1 – Which Object Does self Refer To?

# --------------------------------------------------
# Data to Work With
# --------------------------------------------------

# Player 1:
# Name: Aditya
# Starting score: 0

# Player 2:
# Name: Priya
# Starting score: 0

# Updates, in this order:
# Aditya gains 10 points.
# Priya gains 20 points.
# Aditya gains another 5 points.

# --------------------------------------------------
# Problem Statement
# --------------------------------------------------

# Create a Player class that stores a name and score.
# The score must default to 0.
#
# Include an add_points(points) method that updates the
# current player's stored score and returns the updated score.
#
# Create two player objects and perform the updates above.
# Print each update's result, then display both final scores.
# Keep printing outside the class.
#
# Each player must maintain an independent score.
# Do not create replacement objects or hardcode calculated scores.
#
# Along with your code, explain:
# Which object does self refer to during each method call?
# Why doesn't updating Aditya's score change Priya's score?

# --------------------------------------------------
# Expected Output
# --------------------------------------------------

# Aditya's Updated Score: 10
# Priya's Updated Score: 20
# Aditya's Updated Score: 15
#
# Final Scores:
# Aditya: 15
# Priya: 20

class Player:
    def __init__(self, name, score = 0):
        self.player_name = name
        self.player_score = score

    def add_points(self,points):
        self.player_score += points
        return self.player_score

player_1 = Player("Aditya")
player_2 = Player("Priya")

player_1.add_points(10)
player_2.add_points(20)
print(f"{player_1.player_name}'s Updated Score: {player_1.player_score}")
print(f"{player_2.player_name}'s Updated Score: {player_2.player_score}")
player_1.add_points(5)
print(f"{player_1.player_name}'s Updated Score: {player_1.player_score}")
print()
print(f"Final Scores: ")
print(f"{player_1.player_name}: {player_1.player_score}")
print(f"{player_2.player_name}: {player_2.player_score}")
```

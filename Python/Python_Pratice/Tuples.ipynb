```python
# 📝 Task 1 – Creating and Accessing Tuples
# Create a tuple named employees containing:

# "Rahul", "Priya", "Amit", "Sneha"
# Print:
# First employee
# Last employee
# Total number of employees

# Expected Output
# First Employee: Rahul
# Last Employee: Sneha
# Total Employees: 4

employee = ("Rahul", "Priya", "Amit", "Sneha")
print("First Employee:", employee[0])
print("Last Employee:", employee[-1])
print("Total Employees:",len(employee))
```

```python
#  Task 2 – Tuple Slicing
# Create a tuple:

# (1001, 1002, 1003, 1004, 1005, 1006)
# Print:
# First three IDs
# Last two IDs
# IDs from index 2 to 4

# Expected Output
# First Three: (1001, 1002, 1003)
# Last Two: (1005, 1006)
# Middle IDs: (1003, 1004, 1005)

ID = (1001, 1002, 1003, 1004, 1005, 1006)
print("First three IDs:", ID[0:3])
print("Last Two:",ID[-2:])
print("Middle IDs:",ID[2:5])
```

```python
# Task 3 – Tuple Iteration
# Create a tuple:
# Print each log on a separate line.

# Expected Output
# INFO
# ERROR
# WARNING
# INFO

Logs = ("INFO", "ERROR", "WARNING", "INFO")

for log in Logs:
    print(log)
```

```python
#  Task 4 – Count Occurrences
# Create a tuple:
# Print how many times "Chrome" appears.

# Expected Output
# Chrome Count: 3

browser = ("Chrome", "Firefox", "Chrome", "Edge", "CHROME")
count = 0

for i in browser:
    i = i.lower()      # Convert current tuple element
    if i == "Chrome":
        count += 1

print("Chrome Count:", count)
```

```python
# Task 5 – Find an Index
# Create a tuple:

# ("Python", "SQL", "Spark", "Airflow")
# Print the index of "Spark".

# Expected Output
# Spark Index: 2

language = ("Python", "SQL", "Spark", "Airflow")
for idx,tech in enumerate(language):
    if tech == "Spark":
        print("Spark Index:",idx)
```

```python
# Data Engineering Task – Monthly Sales Report
# A data warehouse stores monthly sales in an immutable tuple because historical data should not be modified.

# Write a program to print:
# Total months
# Total annual sales
# Average monthly sales
# Highest monthly sale
# Lowest monthly sale
# Sales of the first month
# Sales of the last month

# Expected Output
# ========== MONTHLY SALES REPORT ==========
# Total Months: 12
# Total Annual Sales: <calculated value>
# Average Monthly Sales: <calculated value>
# Highest Monthly Sale: 47000
# Lowest Monthly Sale: 25000
# First Month Sales: 25000
# Last Month Sales: 47000
# ==========================================

monthly_sales = (
    25000,
    27000,
    30000,
    28000,
    35000,
    40000,
    38000,
    42000,
    39000,
    41000,
    45000,
    47000
)

print("========== MONTHLY SALES REPORT ==========")
print("Total Months:",len(monthly_sales))
total_sales = 0
average_sales = 0
highest_sale = monthly_sales[0]
lowest_sale = monthly_sales[0]
for sales in monthly_sales:
    total_sales += sales
    if highest_sale < sales:
        highest_sale = sales
    if lowest_sale > sales:
        lowest_sale = sales
average_sales = (total_sales / len(monthly_sales))
print("Total Annual Sales:",total_sales)
print(f"Average Monthly Sales: {average_sales:.1f}")
print("Highest Monthly Sale:",highest_sale)
print("Lowest Monthly Sale:",lowest_sale)
print("First Month Sales:",monthly_sales[0])
print("Last Month Sales:",monthly_sales[-1])
print("==========================================")
```

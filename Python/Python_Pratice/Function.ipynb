```python
# 📝 Task 16 – Utility Function
# Create

# clean_status(status)
# Input:

# " SUCCESS "
# "Failed "
# " pending"
# Expected:

# success
# failed 
# pending

def clean_status(status):
    return status.strip().lower()  # strip() removes leading and trailing whitespace characters from the string, and lower() converts the string to lowercase.

print(clean_status(" SUCCESS "))
print(clean_status("Failed "))
print(clean_status(" pending"))
```

```python
# 📝 Task 18 – Basic Error Prevention
# Create:

# calculate_average(total, count)
# Call:

# 1000,5
# 1000,0
# Expected:

# Average: 200.0
# Average: 0

def calculate_average(total, count):
    if count == 0:
        return 0
    else:
        return total / count  

print(f"Average: {calculate_average(1000,5)}")
print(f"Average: {calculate_average(1000,0)}")         
```

```python
# 📝 Task 19 – Returning Structured Data
# Create:

# pipeline_summary(success, failed)
# Input:

# 950
# 50
# Expected:

# Successful: 950
# Failed: 50
# Total: 1000

def pipeline_summary(success, failed):
    return success + failed , success, failed

total,successful, failed = pipeline_summary(950,50)
print(f"Successful: {successful}")
print(f"Failed: {failed}")
print(f"Total: {total}")
```

```python
# 🎯 Interview Task 2
# Create a function that accepts any number of marks and returns:

# Highest

# Lowest

# Average

# Expected:

# Highest: 95
# Lowest: 72
# Average: 84.2

def marks(*args):
    count = 0
    total = 0
    for score in args:
        count +=1
        total += score
    highest = max(args)
    lowest = min(args)
    average = total/count

    return highest, lowest, average
    
highest, lowest, average = marks (23,46,23,65,45,95)
print(f"Highest: {highest}")
print(f"Lowest: {lowest}")
print(f"Average: {average}")
        
```

```python
# 🏗️ Data Engineering Task 1
# Using:

# daily_records = {
#     "Orders":1250,
#     "Customers":480,
#     "Products":320
# }
# Create:

# print_header()

# total_records()

# highest_dataset()

# print_report()

# Expected:

# ========== DAILY ETL REPORT ==========

# Orders : 1250
# Customers : 480
# Products : 320

# Total Records : 2050
# Largest Dataset : Orders

# ======================================



daily_records = {
    "Orders":1250,
    "Customers":480,
    "Products":320
}  

def header():
    print("========== DAILY ETL REPORT ==========")
    print()
    for key, val in daily_records.items():
        print(f"{key}: {val}")


def total_records():
    total_record = 0
    for val in daily_records.values():
        total_record += val
    print(f"Total Records: {total_record}") 
    largest_dataset = max(daily_records, key=daily_records.get)
    print(f"Largest Dataset: {largest_dataset}")

def print_report():
    header()
    print()
    total_records()
    print("======================================")


print_report()








```

```python
# 🏗️ Data Engineering Task 2
# A data pipeline receives:

# Create separate reusable functions to:

# Count valid records (>0)

# Count invalid records

# Calculate total valid records

# Print the final report

# Expected:

# ========== DATA QUALITY REPORT ==========

# Valid Entries: 4
# Invalid Entries: 3
# Total Valid Records: 1180

# ========================================



records = [120, 0, 350, -10, 500, 0, 210]


def valid_entry(records):
    valid_count = 0
    invalid_count = 0
    total_valid_record = 0
    for record in records:
        if record > 0:
            valid_count += 1
            total_valid_record += record
        else:
            invalid_count +=1
    return valid_count , invalid_count, total_valid_record

valid_entries, invalid_entries, total_valid_entries = valid_entry(records)
print(f"========== DATA QUALITY REPORT ==========\n")
print(f"Valid Entries: {valid_entries}") 
print(f"Invalid Entries: {invalid_entries}") 
print(f"Total Valid Records: {total_valid_entries}\n")
print("========================================")

```

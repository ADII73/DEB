```python
# 📝 Task 5 – Multiple Return Values
# Create:

# calculate(a, b)
# Return:

# Sum

# Difference

# Call:

# 15
# 5
# Expected:

# Sum: 20
# Difference: 10

def calculate(a,b):  
    sum = a + b 
    difference = a - b

    return sum, difference

sum_result, difference_result = calculate(15,5)   # Whenever a function returns multiple values, then we must unpack the function into desired number of variables
print(f"Sum: {sum_result}")
print(f"Differenc: {difference_result}")
```

```python
# Task 1 – Creating and Accessing Dictionary
# Create a dictionary named employee:

# ID : 101
# Name : Rahul
# Department : IT
# Salary : 75000
# Print:
# Employee Name
# Department
# Salary

# Expected Output
# Employee Name: Rahul
# Department: IT
# Salary: 75000

employee = {"ID":101,"Name": "Rahul", "Department" : "IT", "Salary" : 75000}
print(f"Employee Name: {employee["Name"]}")
print(f"Department: {employee["Department"]}")
print(f"Salary: {employee["Salary"]}")
```

```python
# Task 2 – Adding a New Key
# Using this dictionary:

# {
# Name : Priya
# Age : 24
# }
# Add a new key:

# City : Kolkata
# Print the entire dictionary.

# Expected Output
# {'Name': 'Priya', 'Age': 24, 'City': 'Kolkata'}

new_dict = {'Name' : 'Priya', 'Age' : 24}
new_dict['City'] = 'Kolkata'
print(new_dict)
```

```python
# Task 3 – Updating a Value
# Using:

# {
# Product : Laptop
# Price : 65000
# }
# Update the price to:

# 70000
# Print the dictionary.

# Expected Output
# {'Product': 'Laptop', 'Price': 70000}

new_dict = {'Product': 'Laptop', 'Price': 65000}
print(new_dict)
new_dict['Price'] = 70000
print(new_dict)
```

```python
# Task 4 – Removing Data
# Using:

# {
# ID : 101
# Name : Amit
# Department : HR
# }
# Remove the Department key.

# Print the dictionary.

# Expected Output
# {'ID': 101, 'Name': 'Amit'}

dictionary = {'ID': 101, 'Name':'Amit', 'Department':'HR'}
print(dictionary)
del dictionary['Department']
print(dictionary)
```

```python
# Task 5 – Loop Through Dictionary
# Using:

# {
# Python : 95
# SQL : 88
# Spark : 91
# }
# Print each key and value.

# Expected Output
# Python : 95
# SQL : 88
# Spark : 91

dict = {'Python': 95, 'SQL': 88, 'Spark': 91}
for key, val in dict.items():
    print(f"{key} : {val}")
```

```python
# Task 6 – Dictionary Methods
# Using:

# {
# Name : Sneha
# Age : 26
# Department : Finance
# }
# Print:

# All keys

# All values

# All key-value pairs

# Expected Output
# Keys: dict_keys(['Name', 'Age', 'Department'])
# Values: dict_values(['Sneha', 26, 'Finance'])
# Items: dict_items([('Name', 'Sneha'), ('Age', 26), ('Department', 'Finance')])

info = {'Name': 'Sneha', 'Age': 26, 'Department':'Finance'}

print("Keys:", info.keys())
print("Values:", info.values())
print("Items:", info.items())
```

```python
# Task 7 – Membership Check
# Using:

# {
# Python : 95
# SQL : 88
# Spark : 91
# }
# Check whether "SQL" exists.

# Expected Output
# SQL Found

member_check = {'Python': 95, 'SQL': 88, 'Spark': 91}
if 'SQL' in member_check:
    print("SQL Found")
```

```python
# Task 8 – get() Method
# Using:

# {
# Name : Rahul
# Department : IT
# }
# Print:

# Name using get()

# Salary using get()

# Expected Output
# Rahul
# None

info = {'Name': 'Rahul', 'Department': 'IT'}
print(info.get('Name'))
print(info.get('Salary','Not Available'))
print(info['Salary']) # If key not present then it will give error
```

```python
# Task 9 – Nested Dictionary
# Create:

# Employee 101
#     Name : Rahul
#     Salary : 80000

# Employee 102
#     Name : Priya
#     Salary : 90000
# Print:

# Rahul's salary

# Priya's name

# Expected Output
# Rahul Salary: 80000
# Priya Name: Priya

staff = {
    'employee_101' :{'Name': 'Rahul','Salary':80000},
    'employee_102' : {'Name': 'Priya','Salary':90000}
}

print(f"Rahul Salary: {staff['employee_101']['Salary']}")
print(f"Priya Name: {staff['employee_102']['Name']}")
```

```python
# 🎯 Interview Task (included now)
# A company stores employee information as a dictionary.

# {
# 101 : Rahul
# 102 : Priya
# 103 : Amit
# 104 : Sneha
# }
# Without changing the existing data:

# Print the total number of employees.

# Check whether employee ID 102 exists.

# Print the employee name for ID 103.

# Add employee 105 : Neha.

# Print the final dictionary.

# Expected Output
# Total Employees: 4
# Employee Exists
# Employee Name: Amit

# Updated Employee Records:
# {101: 'Rahul', 102: 'Priya', 103: 'Amit', 104: 'Sneha', 105: 'Neha'}

employee = {101: 'Rahul', 102: 'Priya', 103: 'Amit', 104: 'Sneha'}

print(f"Total Employees: {len(employee)}")

if 102 in employee:
    print(f"Employee Exists")

print(f"Employee Name: {employee[103]}")
employee[105] = 'Neha'

print(f"Updated Employee Record:")
print(employee)
```

```python
# 🏢 Data Engineering Task
# A data pipeline has processed records from different files.

# Create this dictionary:


# Generate the report:

# Total customers (dictionary length)

# Total processed records (sum of all values)

# Highest processed records

# Lowest processed records

# Print each customer with its processed record count

# Expected Output
# ========== DAILY PIPELINE REPORT ==========

# Total Customers: 5
# Total Records Processed: 645
# Highest Records: 200
# Lowest Records: 80

# Customer-wise Processing
# A : 120
# B : 95
# C : 150
# D : 80
# E : 200

# ===========================================

customers = {
"A": 120,
"B": 95,
"C": 150,
"D": 80,
"E": 200
}

print("========== DAILY PIPELINE REPORT ==========")
sum = 0
highest = customers["A"]
lowest = customers["A"]
print(f"Total Customers: {len(customers)}")
for val in customers.values():
    sum = sum + val
    if val > highest:
        highest = val
    if val < lowest:
        lowest = val
print("Total Records Processed:",sum)
print("Highest Records:", highest)
print("Lowest Records:", lowest)
print()
print("Customer-wise Processing")
for key, val in customers.items():
    print(f"{key} : {val}")

print("===========================================")
```

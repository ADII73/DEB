```python
# Task 1 – Filter Even Numbers
# Given:


# Use filter() with a lambda to keep only the even numbers.

# Expected Output
# [2, 4, 6, 8]

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

new_numbers = list(filter(lambda num : num % 2 == 0 , numbers))
print(new_numbers)
```

```python
# Task 2 – Filter Positive Numbers
# Given:


# Keep only the positive numbers (> 0).

# Expected Output
# [10, 25, 18]

numbers = [-5, 10, 0, 25, -3, 18]

new_numbers = list(filter(lambda num : num > 0 , numbers))
print(new_numbers)
```

```python
# Task 3 – Filter Long Words
# Given:


# Keep only the words whose length is greater than 4.

# Expected Output
# ['Python', 'Engineering']

words = ["AI", "Python", "Data", "Engineering", "ML"]

new_words = list(filter(lambda String :  len(String) > 4 , words))
print(new_words)
```

```python
# Task 4 – Filter Passing Marks
# Given:


# Keep only the students who scored 40 or more.

# Expected Output
# [40, 72, 90, 55]

marks = [35, 40, 72, 18, 90, 55]

new_numbers = list(filter(lambda num : num >= 40 , marks))
print(new_numbers)
```

```python
# Task 5 – Filter Uppercase Names
# Given:


# Keep only the names that are completely uppercase.

# Hint: Explore the string method .isupper().

# Expected Output
# ['PRIYA', 'JOHN']


names = ["Rahul", "PRIYA", "Amit", "JOHN"]

new_name = list(filter(lambda name : name.isupper()  , names))
print(new_name)
```

```python
# 💼 Interview Tasks
# Interview Task 1 – Filter Multiples of 5
# Given:


# Keep only the numbers divisible by 5.

# Expected Output
# [25, 30, 55, 70]

numbers = [12, 25, 30, 41, 55, 63, 70]

new_numbers = list(filter(lambda num : num % 5== 0 , numbers))
print(new_numbers)
```

```python
# Interview Task 2 – Filter Employees by Salary
# Given:


# Keep only employees whose salary is greater than or equal to 60000.

# Expected Output
# [
#     ("Priya", 65000),
#     ("Sneha", 80000)
# ]

employees = [
    ("Rahul", 50000),
    ("Priya", 65000),
    ("Amit", 45000),
    ("Sneha", 80000)
]

new_numbers = list(filter(lambda num : num [1] >= 60000 , employees))

print(new_numbers) 
    
```

```python
# DE Task 1 – Filter Valid Records
# Given:


# Keep only the records where:

# status == "success"
# Expected Output
# [
#     {"id": 101, "status": "success"},
#     {"id": 103, "status": "success"}
# ]


records = [
    {"id": 101, "status": "success"},
    {"id": 102, "status": "failed"},
    {"id": 103, "status": "success"},
    {"id": 104, "status": "pending"}
]

valid_records = list(filter(lambda record : record["status"] == "success", records))

print(valid_records)
```

```python
# DE Task 2 – Filter High-Value Orders
# Given:


# Keep only the orders where:

# amount >= 2000
# Expected Output
# [
#     {"order_id": 1, "amount": 2500},
#     {"order_id": 3, "amount": 4200}
# ]

orders = [
    {"order_id": 1, "amount": 2500},
    {"order_id": 2, "amount": 800},
    {"order_id": 3, "amount": 4200},
    {"order_id": 4, "amount": 1500}
]

valid_records = list(filter(lambda record : record["amount"] >= 2000, orders))

print(valid_records)
```

```python
# Task 1 – Square Every Number
# Given:

# 
# Use map() with a lambda to square every number.

# Expected Output
# [4, 16, 36, 64]


numbers = [2, 4, 6, 8]
result= map(lambda x : x*x, numbers)
result2= list(map(lambda x : x*x, numbers))
result3 = next(map(lambda x : x*x, numbers))
print(result)
print(result2)
print(result3)
```

```python
# Task 2 – Convert Names to Uppercase
# Given:

# 
# Convert every name to uppercase using map().

# Expected Output
# ['RAHUL', 'PRIYA', 'AMIT', 'SNEHA']

names = ["rahul", "priya", "amit", "sneha"]

new_names = list(map(lambda string : string.upper(), names))
print(new_names)
```

```python
# Task 3 – Add 10 to Every Number
# Given:

# 
# Increase every number by 10.

# Expected Output
# [20, 30, 40, 50]

numbers = [10, 20, 30, 40]

new_number = list(map(lambda num: num + 10, numbers))
print(new_number)
```

```python
# Task 4 – Convert Celsius to Fahrenheit
# Formula:

# F = (C × 9/5) + 32
# Given:

# 
# Expected Output
# [32.0, 68.0, 86.0, 104.0]

temps = [0, 20, 30, 40]

new_temps = list(map(lambda temp: (temp * 9/5) + 32, temps))
print(new_temps)
```

```python
# Task 5 – String Length
# Given:

# 
# Return the length of every word.

# Expected Output
# [6, 4, 11, 2]

words = ["Python", "Data", "Engineering", "AI"]

length = list(map(lambda string : len(string), words))
length = list(map(len, words))
print(length)
```

```python
# Interview Task 1
# Given:

# 
# Increase every salary by 15% using map().

# Expected Output
# [46000.0, 63250.0, 80500.0, 34500.0]

salary = [40000, 55000, 70000, 30000]

new_salary = list(map(lambda num : num + (num * 0.15),salary))
print(new_salary)
```

```python
# Interview Task 2
# Given:

# 
# Use map() to return both:

# Number

# Square

# Output should look like:

# [(1, 1), (2, 4), (3, 9), (4, 16), (5, 25)]
# Hint: A lambda can return a tuple.

numbers = [1, 2, 3, 4, 5]
def squares(num):
        original = num
        square = num*num
        return original, square  # whenever comma between two int it becomes tuple



   
new_number = list(map(squares , numbers))
new_number2 = list(map(lambda num: (num,num*num), numbers))
print(new_number)
print(new_number2)
```

```python
# DE Task 1 – Clean Customer Names
# Given:


# Using map(), clean every name so that:

# Remove extra spaces.

# Convert to title case.

# Expected Output
# ['Rahul', 'Priya', 'Amit', 'Sneha']

customers = [
    " rahul ",
    " PRIYA ",
    " amit ",
    " Sneha "
]

new_names = list(map(lambda string : string.title().strip(), customers))
print(new_names)
```

```python
# DE Task 2 – Convert Records into Dictionaries
# Given:


# Convert them into:

# [
#     {"Name": "Rahul", "Salary": 50000},
#     {"Name": "Priya", "Salary": 65000},
#     {"Name": "Amit", "Salary": 45000}
# ]
# using map()

employees = [
    ("Rahul", 50000),
    ("Priya", 65000),
    ("Amit", 45000)
]

new_employees = list(map(lambda emp : {"Name": emp[0], "Salary": emp[1]} ,employees))
print(new_employees)
```

```python
# Imagine you're building a small ETL transformation.

# Given:


# Perform the following:

# Increase every sale by 18% GST.

# Round the result to 2 decimal places.

# Print the final list.

# Expected Output
# [1770.0, 2950.0, 2124.0, 3776.0, 3304.0]


sales = [1500, 2500, 1800, 3200, 2800]

new_sales = list(map(lambda a : round(a + (a * 0.18), 2), sales))
print(new_sales)
```

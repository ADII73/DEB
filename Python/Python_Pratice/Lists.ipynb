```python
# Task 1 – Employee Management
# Perform the following:
# Add "Rohan" to the end.
# Insert "Neha" at index 2.
# Remove "Priya".
# Print the final list.

# Expected Output
# ['Rahul', 'Neha', 'Amit', 'Sneha', 'Rohan']


employees = ["Rahul", "Priya", "Amit", "Sneha"]

employees.append("Rohan")
print(employees)
employees.insert(2,"Neha")
print(employees)
employees.remove("Priya")
print(employees)
```

```python
# Task 3 – Student Marks
# Print each mark with its position.

# Expected Output
# Student 1 : 75
# Student 2 : 82
# Student 3 : 67
# Student 4 : 91
# Student 5 : 88

marks = [75, 82, 67, 91, 88]

for index, score in enumerate(marks):
    print(f"Student {index+1} : {score}")
```

```python
# Task 4 – Remove Duplicate IDs
# Without using set():
# Create a new list containing only unique IDs.
# Preserve the original order.

# Expected Output
# [101, 102, 103, 104, 105]

employee_ids = [101, 102, 103, 102, 104, 101, 105]
unique_ids = []

for unique in employee_ids:
    if unique not in unique_ids:
        unique_ids.append(unique)
print(unique_ids)

# print(set(employee_ids))

```

```python
# Task 5 – Search Operation
# Check whether "Kolkata" exists.

# Expected Output
# City Found
# If it doesn't exist:
# City Not Found

cities = ["Delhi", "Mumbai", "Kolkata", "Chennai", "Bangalore"]

if "Kolkata" in cities:
    print("City Found")
else:
    print("City Not Found")
```

```python
# Task 6 – Count Occurrences
# Without using .count():

# Print:

# Credit : 3
# Debit : 2

transactions = ["Credit", "Debit", "Credit", "Credit", "Debit"]

credit_count = 0
debit_count = 0
for card in transactions:
    if card.lower() == "credit":
        credit_count += 1
    elif card.lower() == "debit":
        debit_count += 1

print(credit_count)
print(debit_count)
```

```python
# Task 7 – Reverse a List
# Print the list in reverse order.

# Expected Output
# [50, 40, 30, 20, 10]

numbers = [10, 20, 30, 40, 50]
new = []

print(numbers[::-1])
numbers.reverse()
print(numbers)

for num in numbers:  
    new.insert(0, num)

print(new)
```

```python
# Task 8 – Largest & Smallest
# Without using max() or min():

# Print

# Highest Temperature : 35
# Lowest Temperature : 24

temperatures = [31, 28, 35, 24, 30, 33]

min = temperatures[0]
max = temperatures[0]

for temp in temperatures:
    if min > temp:
        min = temp
    elif max < temp:
        max = temp

print(min)
print(max)
```

```python
# Data Engineering Task
# Scenario
# You are a Junior Data Engineer at an online shopping company.

# The sales team sends today's order values.

# Your manager asks you to prepare the daily report.

# Requirements
# Generate a report containing:
# Total number of orders
# Total revenue
# Average order value
# Highest order
# Lowest order
# Number of orders greater than ₹3000
# Number of orders less than ₹1000
# Create two new lists:
# high_value_orders
# Contains orders greater than ₹3000.
# low_value_orders
# Contains orders less than ₹1000.
# Finally print:

# ========== DAILY ORDER REPORT ==========

# Total Orders :
# Total Revenue :
# Average Order Value :
# Highest Order :
# Lowest Order :

# Orders > 3000 :
# Orders < 1000 :

# High Value Orders :
# [...]

# Low Value Orders :
# [...]

# ========================================

orders = [2500, 1800, 3200, 450, 5100, 700, 2900, 1500, 6000, 900]

total_order = 0
total_revenue = 0
min = orders[0]
max = orders[0]
orders_less_3000 = 0
orders_more_1000 = 0
high_value_orders = []
low_value_orders = []

for order in orders:
    total_order += 1
    total_revenue += order
    if min > order:
        min = order
    elif max < order:
        max = order
    if order > 3000:
        orders_less_3000 += 1
        high_value_orders.append(order)
    elif order < 1000:
        orders_more_1000 +=1
        low_value_orders.append(order)

    


average_order_value = total_revenue / total_order

print("========== DAILY ORDER REPORT ==========")

print(f"Total Orders : {total_order}")
print(f"Total Revenue : {total_revenue}")
print(f"Average Order Value : {average_order_value}")
print(f"Highest Order : {max}")
print(f"Lowest Order : {min}")
print()
print(f"Orders < 3000 : {orders_less_3000}")
print(f"Orders > 1000 : {orders_more_1000}")
print()
print("High Value Orders :")
print(high_value_orders)
print()
print("Low Value Orders :")
print(low_value_orders)

print("========================================")




```

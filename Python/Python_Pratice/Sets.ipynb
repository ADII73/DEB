```python
# Task 1 – Remove Duplicate Cities
# Scenario
# A travel company receives city names with duplicates.

# Requirements
# Convert the list into a set.
# Print the set.
# Print the number of unique cities.
# Expected Output

# Unique Cities:
# {'Delhi', 'Mumbai', 'Kolkata', 'Chennai'}
# Total Unique Cities:
# 4


cities = [
    "Delhi",
    "Mumbai",
    "Delhi",
    "Kolkata",
    "Mumbai",
    "Chennai"
]
count = 0
unique_cities= set(cities)
print("Unique Cities:")
print(unique_cities)

for city in unique_cities:
    count = count + 1

print(count)
print(len(unique_cities))
```

```python
# Task 4 – Students in Two Courses
# Print:
# Students enrolled in either course.
# Students enrolled in both courses.
# Expected Output
# All Students:
# {...}

# Common Students:
# {'Amit', 'Priya'}

python_students = {"Rahul", "Amit", "Sneha", "Priya"}
sql_students = {"Amit", "Priya", "Karan", "Neha"}

all_students = python_students.union(sql_students)
print("All Students: ",all_students)

common_students = python_students.intersection(sql_students)
print("Common Students: ",common_students)

    
```

```python
# Task 5 – Exclusive Python Students
# Using the same sets above,

# Print students enrolled only in Python.

# Expected Output
# {'Rahul', 'Sneha'}

only_python = all_students.difference(sql_students)
print(only_python)
```

```python
# Task 6 – Exclusive SQL Students
# Using the same sets,

# Print students enrolled only in SQL.

# Expected Output
# {'Karan', 'Neha'}

sql_student = all_students.difference(python_students)
print(sql_student)
```

```python
# Task 7 – Different Students
# Using the same sets,

# Print students who are in exactly one of the courses.

# Expected Output
# {'Rahul', 'Sneha', 'Karan', 'Neha'}

only_one_course = python_students.symmetric_difference(sql_students)
print(only_one_course)
```

```python
# Data Engineering Task
# Scenario
# You are a Junior Data Engineer at an online learning platform.

# Two systems send today's user IDs.

# Your manager asks you to prepare a report.

# Requirements
# Print:
# Total unique users across both platforms.
# Users who logged in from both platforms.
# Users who logged in only from the website.
# Users who logged in only from the mobile app.
# Users who logged in from exactly one platform.
# Finally print the report in this format:

# ========== USER LOGIN REPORT ==========
# All Users:
# {...}
# Common Users:
# {...}
# Web Only Users:
# {...}
# Mobile Only Users:
# {...}
# Exactly One Platform:
# {...}
# =======================================


web_users = {101, 102, 103, 104, 105, 106}
mobile_users = { 104, 105, 106, 107, 108, 109}

print("========== USER LOGIN REPORT ==========")
print("All Users:")
unique_users = web_users.union(mobile_users)
print(unique_users)

print("Common Users:")
common_users = web_users.intersection(mobile_users)
print(common_users)

web_only_users = web_users.difference(mobile_users)
print("Web Only Users:")
print(web_only_users)

print("Mobile Only Users:")
mobile_only_users = mobile_users.difference(web_users)
print(mobile_only_users) 

print("Exactly One Platform:")
exactly_one_platform = web_users.symmetric_difference(mobile_users)
print(exactly_one_platform)
```

```python
# Task 1 – Create a lambda function to calculate the square of a number.

# Expected Output

# Square: 64

sqr = lambda a : a*a
print(f"Square:{sqr(8)}") 
```

```python
# Task 2 – Create a lambda function to add two numbers.

# Expected Output

# Sum: 35
add = lambda a,b : a+b
print(f"Sum: {add(8,27)}") 
```

```python
# Task 3 – Create a lambda function to find the larger of two numbers.

# Expected Output

# Largest: 20

check = lambda a,b : a if a>b else b
print(f"Largest: {check(8,20)}") 
```

```python
# Task 4 – Create a lambda function to check if a number is even.

# Expected Output

# True

even_odd = lambda a : a % 2 == 0
print(f"{even_odd(6)}")
```

```python
# Task 5 – Create a lambda function that converts a string to uppercase.

# Expected Output

# DATA ENGINEERING

conversion = lambda string : string.upper()
print(f"{conversion('data engineering')}")
```

```python
# Interview Task 1

# Create a lambda function that returns "Pass" if marks are greater than or equal to 40, otherwise "Fail".

# Expected Output

# Pass
# Fail

result = lambda score : "Pass" if score >= 40 else "Fail"
print(f"{result(40)}")
print(f"{result(39)}")
```

```python
# Interview Task 2

# Create a lambda function with three parameters that returns the largest number.

# Expected Output

# Largest: 50


largest = lambda a, b, c : a if a > b and a > c else  b if b > a and b > c else c
print(f"Largest: {largest(10,20,50)}")
```

```python
# DE Task 1
# Given:


# Create a lambda function that returns True only if the status is "success".

# Expected Output

# Valid Record: True
record = {
    "id":101,
    "status":"success"
}
records = lambda record : record["status"] == "success"
print(f"Valid Record: {records(record)}")
```

```python
# DE Task 2
# Given:

# salary = 50000
# Create a lambda function that increases the salary by 10%.

# Expected Output

# Updated Salary: 55000

increase_salary = lambda salary : salary  + int(salary * 0.10)
print(f"Updated Salary: {increase_salary(50000)}")
```

```python
# 🚀 Mini Challenge
# Without using def, create a program that:

# Calculates the square of a number.

# Checks whether it is even or odd.

# Prints "Large Number" if the square is greater than 100, otherwise "Small Number".

# This combines multiple lambda concepts into one exercise.

square = lambda num: num * num
even_odd = lambda sq: "Even" if sq % 2 == 0 else "Odd"
size = lambda sq: "Large Number" if sq >= 100 else "Small Number"

num = 10

sq = square(num)

print("Square:", sq)
print("Type:", even_odd(sq))
print(size(sq))
```

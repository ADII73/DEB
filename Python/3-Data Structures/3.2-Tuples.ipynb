```python
# Tuple is similar to list except tuple is immutable(values cannot be changed)

mytuple=(1,2,3,4,4,4,4,5,6)
print(mytuple)
print(type(mytuple)) 
list(mytuple) # tuple to list

mixed_tuple=(1,2.4,"Adityya",True)
print(mixed_tuple)


# accessing and slicing will be just like lists
```

```python
# Tuples operation

concat=mytuple+mixed_tuple # concatination of two tuples
print(concat)

mytuple*3 # repeat tuples together
mixed_tuple*3
```

```python
# Immutable nature
# mytuple[0]=10 # values cannot be changed like list

  
```

```python
# methods

print(mytuple.count(4)) # count of given element
print(mytuple.index(5)) # index of given element

```

```python
# Packing 
packed_tuple = 2,3.14,"Aditya",True   # without creating tuple, elements of different type can be packed together 
print(packed_tuple)
print()


# Unpacking
a,b,c,d=packed_tuple # to unpack create as number of variable = number of elements in packed_tuple  
print(a)
print(b)
print(c)
print(d)
print()
# Unpacking with star
numbers=(1,2,3,4,56,7)
first,*middle, scndlast, last=numbers # *variable will take the all remaining elments excluding front and back variables
print(first)
print(middle)
print(scndlast)
print(last)
```

```python
# Nested tuples and nested lists

nestedtuples=((0,1,3),("Aditya",3.14,True),("Routh",6,0))
print(nestedtuples[0]) # will give the one tuple inside tuple
print(nestedtuples[1][1]) # will give the element inside the tuple inside nested tuple

# there can be tuples inside list and list inside tuples just access similarly

# iterating over nested tuples
for sub_tuples in nestedtuples:
    for items in sub_tuples:
        print(items,end=" ")

    print()

# end is parameter of print statement, it decides what to print at the end
```

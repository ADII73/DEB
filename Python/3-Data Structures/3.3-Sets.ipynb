```python
# sets are built in data type used to store collections of unique items
# Unordered and do not allow duplicate elements(use for eliminating duplicate elements)

my_set={11,22,33,44,55,5,5,5,5} # only will give unique records
print(my_set)

my_empty_set=set() # empty set

```

```python
my_set.add(1000) # adding element
print(my_set)
my_set.add(1000) # it wont give any error it will simply ignore the duplicate value

my_set.remove(1000) # it will take value
print(my_set)
# if we .remove() an element which is not in the set, it will throw error
my_set.discard(100) # use discard() instead of remove when dont want to show error
print(my_set)

# pop element
poped_element=my_set.pop() # in set pop() will take element from first not like in lists(from last)
print(poped_element)
print(my_set) 

# my_set.clear() it will clear
```

```python
# membership test(to check if any element is present in set or not)
print( 3 in my_set)  # in will check if present or not , it will give true or false
print( 22 in my_set)

```

```python
# mathematical operations

set1={1,2,3,4,5,6,6}
set2={4,5,6,7,8,9,10}

union_set=set1.union(set2) # it will combine both the sets and give data excluding duplicates
print(union_set)

intersection_set=set1.intersection(set2)
print(intersection_set) # it will give values that are common in both the sets

set1.intersection_update(set2) # it will give intersect record (common) and will update in set1 if set2 first then update in set2
print(set1)

```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9,10}

print(set1.difference(set2)) # from set1 minus elements that are present in set2
print(set2.difference(set1)) # from set2 minus elements that are present in set1
print(set1.symmetric_difference(set2)) # from both sets remove common elements and give unique elements

```

```python
set1={1,2,3,4,5}
set2={3,4,5}

print(set1.issubset(set2))  # set1 value is present in set2 or not
print(set2.issubset(set1))  # set2 value is present in set1 or not
print(set1.issuperset(set2)) # instead of reversing sets use issuperset 

```

```python
# count unique words in sentence
sentence="My name is aditya routh, aditya is very good"
words=sentence.split()
print(words)
unique=set(words)
print(unique)
print(len(unique))
```

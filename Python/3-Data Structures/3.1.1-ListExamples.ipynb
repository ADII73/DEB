```python
# Create a to do list
my_list=["wakeup","do exercise","make bed","go to office"]

# Adding to task
my_list.append("Eat food") # will add at end
my_list.insert(2,"Bathing")
print(my_list )

#removing
my_list.remove("do exercise")
print(my_list )

# check if task in list

if "do exercise" in my_list:
    print("to do")
else:  
    print("done")

# check remaining actions
print("To do list:")
for task in my_list:
    print(f"-{task}")
```

```python
# organizing student grades
grades=[85,92,78,99]

# add a new grade
grades.append(89)

# calculate the avearge grade
average_grade=sum(grades)/len(grades)
print(f"The average grade of class is {average_grade:.2f}")

# highest and lowest
max_score=max(grades)
min_score=min(grades)

print(f"Highest score {max_score}")
print(f"Lowest score {min_score}")
```

```python
# inventory
inventory=["apples","banana","kiwi","guava","lemon"]

#add in inventory
inventory.append("watermelon")

# remove/buy 
inventory.remove("banana")

# check if in stock or not
item="orange"
if item in inventory:
    print(f"{item} is in stock")
else:   
    print(f"{item} is not in stock")

print(f"These are the items in inventory")
for item in inventory:
    print(f"-{item}")
```

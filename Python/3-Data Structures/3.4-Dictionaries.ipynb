```python
student={"name":"Aditya", "Age": 22, "Gender":"Male"}
print(student)
print(type(student))
# same key cannot be there, if given then the last occurence of the same key will give its value to first occurance of its same key
new_student={"name":"Aditya", "Age": 22, "Gender":"Male", "name":122}
print(new_student)
```

```python
students={"name":"Aditya", "Age": 22, "Gender":"Male", "grade":"A"}
print(students["grade"])
print(students["name"])
# print(students["phone"]) # this will give error if key not present

#  accessing element by get()
print(students.get("grade"))
print(students.get("phone","Not Available")) # this will not give error and if not present in the dict, can give alternate value
```

```python
# Modifying Key value pair
students["height"]=33  # Adding new key value pair
print(students)

students["Gender"]="Female" # Updating existing key value
print(students)

# Deleting key value pair 
del students["height"]
print(students)
```

```python
# Dictionary methods

keys=students.keys() # This will give all the keys 
print(keys)

values=students.values() # This will give all the values 
print(values) 

items=students.items() 
print(items) # This will give entire dict in pairs , list of tuples

print(students) # this will give entire dictionary, mainly used for specific value of an key
```

```python
# Shallow copying

student_copy=students
print(students)
print(student_copy)


```

```python
student_copy["Age"]=92
print(student_copy)  # Changing value of any of the two dictionary, it will refelct in both, i.e. shallow copy
print(students)

```

```python
student_copy1=student.copy()
student_copy["Age"]=9998  # Change will only happen in one dictionary
print(students)
print(student_copy1)
```

```python
# Iterating over dictionary

for key in students.keys():
    print(key)
print()
for value in students.values():
    print(value)
print()
for key,value in students.items():
    print(f"{key}:{value}")
print()
for key,value in students.items():
    print(f"{key}:{value}")
```

```python
#Nested dictionaries

student={
    "student1":{"name":"Aditya", "Age": 2, "Hobby": "Nothing"},
    "student2":{"name":"Adityaa", "Age": 22, "Hobby": "Nothingg"},
    "student3":{"name":"Adityaaa", "Age": 222, "Hobby": "Nothinggg"},
    "student4":{"name":"Adityaaaa", "Age": 2222, "Hobby": "Nothingggg"},
    "student5":{"name":"Adityaaaaa", "Age": 22222, "Hobby": "Nothinggggg"}
}
print(student)

#Accessing nested dictionaries elements
print(student["student2"]["name"])
print(student["student2"]["Age"])
```

```python
# Iterating over nested dict

for student_id,student_info in student.items():
    print(f"{student_id}:{student_info}")

    for key, value in student_info.items():
        print(f"{key},{value}")

    print()
```

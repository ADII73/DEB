```python
class Car:    # Single  Inheritance

    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The car runs on {self.enginetype}")


class Tesla(Car):

    def __init__(self,windows,doors,enginetype,isselfdrive):

        super().__init__(windows,doors,enginetype)
        self.isselfdrive = isselfdrive

    def selfdrive(self):
        print(f"The car can self drive? {self.isselfdrive}")

car1 = Car(4,5,"Petrol")
print(car1.windows,car1.doors,car1.enginetype)
car1.drive()


car2 = Tesla(2,4,"Electric",True)
print(car2.windows, car2.doors, car2.enginetype, car2.isselfdrive)
car2.selfdrive()
car2.drive()
```

```python
# Multiple Inheritance

class Animal:
    def __init__(self, name):
        self.name = name

    def speak1(self):
        print(f"Subclass must implement this method")

class Pet:
    def __init__(self, owner):
        self.owner = owner

class Dog(Animal,Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} says woof"


dog = Dog("Buddy", "Aditya")
print(dog.speak())
print(f"owner: {dog.owner}")
dog.speak1()

    
```
